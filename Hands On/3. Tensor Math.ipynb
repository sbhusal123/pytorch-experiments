## 1. Tensor Math

```python
import torch
```

**Tensor Initiation From Uniform Distribution:**

First, let’s instantiate two random tensors, one with uniform distribution in the range [–1, 1) and the
other with a standard normal distribution:

```python
torch.manual_seed(1)
t1 = 2 * torch.rand(5, 2) - 1
t2 = torch.normal(mean=0, std=1, size=(5, 2))

print("Random standard normal distribution:\n",t1)
print("\nFrom distribution where, mean=0 std=1:\n",t1)

print("\n\nMean (t1) => ", torch.mean(t1))
print("Mean (t2) => ", torch.mean(t2))

print("\n\nStandard Dev (t1) => ", torch.std(t1))
print("Standard Dev (t2) => ", torch.std(t2))
```

Now we can calculate:
- torch.mean(<torch_tensor>, axios=..)
- torch.sum(<torch_tensor>, axios=..)

**Examples**
- Mean along different axis

```python
t1 = torch.tensor([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

t2 = torch.mean(t1, axis=0, dtype=float)
```

**Addition:**

```python
x = torch.tensor([1, 2, 3])

y = torch.tensor([4, 5, 6])

z = x + y

print(z)
print(torch.add(x, y))
```

**Subtraction**

```python
x = torch.tensor([4, 5, 6])
y = torch.tensor([1, 2, 3])

z = x - y

print(z)

print(torch.sub(x, y))
```

**Division:**
- Resulting Tensor in float

```python
x = torch.tensor([4, 6, 10])

y = torch.tensor([2, 3, 5])


# y can be a tensor
print("Div two tensor: ", torch.true_divide(x, y))

# y can be scalar value too
print("Divice tensor by scalar: ",torch.true_divide(x, 2))
```

**Exponential:**

```python
x = torch.tensor([2, 4, 6])

print("x pow 2:", x ** 2)

print("x pow 2", x.pow(2))
```

# 2. Matrix and Vector

**Matrix Multiplication:**
- Note that, `x*y` is a point wise matrix multiplication

```python
x = torch.rand((2, 5)) # 2 by 5
y = torch.rand((5, 3)) # 5 by 3

z = x.mm(y) ## 2 by 3

z
```

**Point Wise Matrix Multiplication:**
- Multiplies each corresponding element

```python
x = torch.tensor([[1, 2, 3],
                  [4, 5, 6],
                  [6, 7, 9]])

y = torch.tensor([[3, 2, 1],
                  [8, 5, 7],
                  [4, 1, 2]])

x * y
```

**Matrix Exponentiation:**
- $A^3=A*A*A$

```python
a = torch.rand((5, 5))
a.matrix_power(3)
```

**Vector Dot Product:**

```python
x = torch.tensor([1, 2, 3])
y = torch.tensor([4, 5, 6])

z = x.dot(y)

z
```

## 3. Batch Matrix Multiplication:
- Matrix multiplication in batch

```python
batch = 2
n = 2
m = 3
p = 4


t1 = torch.rand(batch, n,m) # two batch of  2x3 matrix
t2 = torch.rand(batch, m,p) # two batch  3x4 matrix

t3 = t1.bmm(t2) # two batch of  2x4 matrix

print("t1: ", t1)
print("\nt2: ", t2)
print("\nt3: ", t3)

print("\nSize => ",t3.shape)
```

# 4. Broadcasting:

**Given tensors:**

$$
\mathbf{X} = \begin{bmatrix}
10 & 20 & 30 \\
40 & 50 & 60 \\
60 & 70 & 80
\end{bmatrix}
\quad
\mathbf{Y} = \begin{bmatrix}
10 & 20 & 30
\end{bmatrix}
$$

**Broadcasting Mechanism:**

$$
\mathbf{Y}_{\text{broadcasted}} = \begin{bmatrix}
10 & 20 & 30 \\
10 & 20 & 30 \\
10 & 20 & 30
\end{bmatrix}
$$

**Element-wise Subtraction:**

$$
\mathbf{X} - \mathbf{Y}_{\text{broadcasted}} =
\begin{bmatrix}
10 - 10 & 20 - 20 & 30 - 30 \\
40 - 10 & 50 - 20 & 60 - 30 \\
60 - 10 & 70 - 20 & 80 - 30
\end{bmatrix}
=
\begin{bmatrix}
0 & 0 & 0 \\
30 & 30 & 30 \\
50 & 50 & 50
\end{bmatrix}
$$


```python
x = torch.tensor([[10, 20, 30],
                  [40, 50, 60],
                  [60, 70, 80]])

y = torch.tensor([[10, 20, 30]])

x - y
```

```python
x + y
```

```python
x * y
```
