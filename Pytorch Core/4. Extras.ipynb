```python
import torch
```

## 1. Random Number from Normal Distribution

```python
# Create a random tensor of shape (3, 4)

x = torch.randn(3, 4)
print(x)
```

## 2. View

What it does: Reshapes a tensor without changing its data. It's the most common way to flatten or expand dimensions. Rule: The total number of elements must remain the same ($B×T×C$ must equal the new dimensions multiplied).


```python
# generate random tensor of shape (1,2,3) in normal distribution
x = torch.randn(1, 2, 3)

# Reshape tensor
y = x.view(2, 3)

print("X = ", x)
print("\n\nX shape = ",x.shape)

print("\n\nY = ", y)
print("\nY shape = ",x.shape)
```

## 3. Splitting a Tensor

```python
# generate random tensor of shape (2,4)
x = torch.randn(2, 4)
print("\nx = ",x)
print("\nx shape = ",x.shape)

# Split tensor into 2 parts along dimension 1 i.e. columns
y = x.split(2, dim=1)
print("\ny = ",y)
print("\ny shape = ",[i.shape for i in y])
```

## 4. Unsqueezing a tensor

```python
pos = torch.arange(0, 10) # Shape: (10)
print("pos: ",pos)
print(f"Pos: {pos.shape}")


pos_batch = pos.unsqueeze(0) # Shape: (1, 10)
print("\n\npos_batch: ",pos_batch)
print(f"Pos Batch: {pos_batch.shape}")
```

## 5. Masked Filling

```python
x = torch.randn(3)

masked = torch.tensor([True, False, True])

print(x)
print(masked)

print(x.masked_fill(masked, -999))
```
