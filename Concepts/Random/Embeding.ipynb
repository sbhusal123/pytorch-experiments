# Embedding 

- **learned vector representation** of discrete items such as words, characters, or tokens.

- Words with **similar meanings or usage tend to have close embeddings in vector space.**


### How Embeddings Work Mathematically

- Suppose vocabulary size is $V$, and embedding dimension is $ d $.

- Embedding matrix $ E $ has shape $ V \times d $. Each row $ E_i $ corresponds to the embedding vector for the $ i^{th} $ word.

- Given a one-hot vector $ x \in \mathbb{R}^V $ for a word (1 at the index of the word, 0 elsewhere), embedding lookup is:

    $$ e = E^T x $$

Because $ x $ is one-hot, this operation just selects the row $ E_i $ corresponding to the word index $ i $.

In practice, embeddings are learned parameters optimized during model training.


### Vocabulary and Vocabulary Size

- **Vocabulary** refers to the set of all unique tokens (words, subwords, or characters) that a model recognizes and processes.  
- Each token in the vocabulary is assigned a unique integer index used to convert text into sequences of numbers for model input.  
- **Vocabulary size** (\( V \)) is the total number of unique tokens in the vocabulary.  

For example, if the vocabulary consists of the words:  
`["cat", "dog", "bird", "fish"]`  
then the vocabulary size \( V = 4 \).  

- A larger vocabulary allows the model to understand more words but increases computational cost.  
- A smaller vocabulary reduces model size but may require tokenizing rare words into subwords or characters.


## 1. One-Hot Encoding (Baseline):


Mathematics
- Words in vocabulary = $\set{"Apple", "Queen", "King"}$
- Vocabulary size: $((V))$

- Each word is represented as a vector: $x \in \{0, 1\}^{v}$  with exactly one 1, all other entries 0.

```python
import torch

vocab = ["cat", "dog", "mouse", "bird", "fish"]

# Vocabulary size
vocab_size = len(vocab)

# Create zero tensor
one_hot = torch.zeros(vocab_size, dtype=torch.float32)

# Example word to convert to one-hot encoding
word = "bird"
index = vocab.index(word)
one_hot[index] = 1.0
print(one_hot)
```

Example
For vocab = [“cat”, “dog”, “mouse”, “bird”, “fish”], “bird” → [0, 0, 0, 1, 0]

## 2. Embeeding (DenseVector)

**Mathematics**
- Embedding matrix $E \in \real^{v \times d}$
- Word index $i$ selects embeding vector: $E_i \in \real^{d}$
- Learned through training.
 

```python
import torch
import torch.nn as nn

# Define a small vocabulary and embedding dimension
vocab = ["cat", "dog", "mouse", "bird", "fish"]
vocab_size = len(vocab)
embedding_dim = 4  # You can pick any dimension

# Create the embedding layer
embedding = nn.Embedding(vocab_size, embedding_dim)

# Example words to get embeddings for
words = ["bird", "cat", "fish"]

# Convert words to their indices in the vocabulary
word_indices = torch.tensor([vocab.index(word) for word in words])

# Get embeddings for these word indices
embedded_vectors = embedding(word_indices)

print("Words:", words)
print("Indices:", word_indices)
print("Embeddings:\n", embedded_vectors)
```
