## 1. Basic Terminilogy and Concepts

---
### i. Experiment:
- On operation which can produce some well defined outcome or cases.

### ii. Outcomes:
- Outcome or a result of an experiment.

### iii. Random Experiment:
- If in each trial of an experiment performed under same conditions, the outcomes are not unique, but may be possibile (**probability**) outcomes of unique outcomes.
- Such an experiment is called as Random Experiment.


### iv. Sample Space and Exhaustive Cases:

- **Sample Space:** Set (set itself is unique) of all possible outcomes in a random experiment. Denoted by **S**.
- **Exhaustive Cases:** No of items in a sample space set.

For example, **sample space** of rolling dice:
$$
S = \{1, 2, 3, 4, 5, 6\} 
$$

**Exhaustive Cases** = 6

---

### v. Events:
- Something that happens (outcomes or combination of outcomes of an experiment)
- It is a **subset of a sample space.**

For Example, when a dice is thrown:

$$
S = \{1, 2, 3, 4, 5, 6 \}
$$


If i thrown a dice now, event of getting even number is given by:

$$
E=\{ 2, 4, 6 \}
$$

Here, $E \subset S$

- **Sure Event:** If the occurance of event is certain. **Example:** Probability of getting number greater than 6 in a throw of a die.
- **Impossible Event:** Occurance of event is impossible. **Example:** Probability of getting number lesser than 6 in a throw of a die.


### vi. Simple Event And Compound Event:

- Event is called **simple event** if it corresponds to a single possible outcomes. 
    - In a throw of die getting a 3 is simple event. $E=\{3\}$
- Event is called **compound event** if it corresponds to joint occurance of two or more events at the same time.
    - In a throw of a die getting even number is Compound Event. $E=\{ 2, 4, 6 \}$

### vii. Equally likely events (or cases): (Relative Events)

- Two or more events are **eqlauuy likely** if they have equal chance of occuring.
- In a throw of a die, probability of getting 1 or 2 or 3 or 4 or 5 or 6 is equally likely.
- $P(E_1) =P(E_2)=P(E_3)...=P(E_n)$

### viii. Mutually Exclusive Events (Non Relative Events)
- Two or more events are mutually exclusive if they cannot occur simultaneously.
- Probability of getting even ($E_1$) or odd ($E_2$) in a two or a die are mutually exclusive events.
- Here, $E_1 \cap E_2= \phi$ i.e. null set.

### ix. Favourable Events Or Cases:
- Outcome of a random experiment taken in favour of happening an event.
- Example: If we **toss two coins simultaneously**  **favourable cases of getting head and tail alternatvely** are: $E=\{ HT, TH \}$



---

## 2. Classical (Mathematical) Definition Of Probability:

If in a random experiments there are 
- **n** mutually exclusive. exhaustive and equally likely cases
- **m** of them are favourable to event A

Then probability of happening of event denoted by $P(A)$ is defined by:

$$
P(A) = \frac{\text{Number of favourable cases}}{\text{Total Number of outcomes}} = \frac{m}{n}
$$

**Notes:**
- In above definition m and n are always positive integer.
- If event A is **sure event**, then P(A) = 1 and if event is **impossible**, then P(A)=0.
- Probability of non occurance is $P(\bar{A})$ such that, $P(\bar{A}) + P(A) = 1$

## 3. Numerical Examples

**1. If a normal die is rolled, find the probability that the number on the upperface is:**

- i. 4
- ii. Not 4
- iii. Prime number

If a normal die is thrown, then the sample space is; S = $\{ 1, 2, 3, 4, 5, 6 \}$

Therfore, total number of possible outcome; $n=6$

**(i). Number of favorable cases of getting 4 is = 1**

Therfore probability of getting 4 = $p(\text{getting 4}) = \frac{1}{6}$

**(ii. ) probability of not getting 4** = $1 - p(\text{getting 4}) = 1 -  \frac{1}{6} = \frac{5}{6}$

**(iii.).** Probability of getting prime $S= \{2, 3, 5 \} = \frac{3}{6} = \frac{1}{2}$ 

---

## 4. Statistical (Emperical) Definition Of Probability:

If an **event A occurs m times in n repetitions of a random experiment**, then, the value of relative frequency $\frac{m}{n}$ **when n is indefinitely large is called the probability of the event A.**

Symbolically,

$$
P(A) = \lim_{n \to \infty} \frac{m}{n}, \text{ provided limit exists}
$$

This is called relative frequency because the number of successful experments is being related to total number of experiments.

**Limitations:**
- A large number of trials cannot be conducted under similar and homonogenous conditions.
- The value of $\frac{m}{n}$ goes on changing as n is made larger and larger so its not possible to get exact value of probability.


## 5. Examples

**1. A bag contains 8 red and 5 white balls. Three are drawn at random. Find the probability that:**

- i. All the three balls are white.
- ii. Two balls are red and one ball is white.

**Answer**

- Number of red balls = 8

- Number of white balls = 5

- Total Number of Balls = 8 + 5 = 13

- If three balls are drawn at random; n = C(13, 3) = $\frac{13!}{(13-3)! 3!}$

$$ n = C(13, 3) =  \frac{13!}{(13-3)! 3!}  = 286 $$

**i. Let $E_1$ be the event of getting 3 white balls**, then number of favourable cases of getting event $m_1=C(5, 3)$

$$
m_1 = C(5, 3) = \frac{5!}{(5-3)!3!} = \frac{5!}{2!3!} = 10
$$

Thus, 

$$
P(E_1) = \frac{m_1}{n} = \frac{10}{286} = \frac{5}{143}
$$

**ii. Let $E_2 $ be the event of getting 2 red and 1 white balls,** then number of favourable cases of getting event:

$$
m_2 = C(8, 2) * C(5, 1) = \frac{8!}{(8-2)!2!} * \frac{5!}{(5-1)!1!} = 140
$$

$$
P(E_2) = \frac{m_2}{n} = \frac{140}{286} = \frac{70}{143}
$$
